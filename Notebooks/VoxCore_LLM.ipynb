```python
# ============================================================
# Cell 2: Imports & Embedding Model
# ============================================================

from llama_index.core import (
    Settings,
    VectorStoreIndex,
    SimpleDirectoryReader
)

from llama_index.core.node_parser import SentenceSplitter

from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# ------------------------------------------------------------
# Embedding Model
# ------------------------------------------------------------

embed_model = HuggingFaceEmbedding(
    model_name="BAAI/bge-base-en-v1.5"
)

Settings.embed_model = embed_model

print("✅ Embedding Model Loaded Successfully!")
```

```python
# ============================================================
# Cell 3: Load Transcript
# ============================================================

from google.colab import files

# Upload transcript file
uploaded = files.upload()

# Load the transcript
documents = SimpleDirectoryReader(
    input_files=["transcript.txt"]
).load_data()

print(f"✅ Loaded {len(documents)} document(s)")
print("\nPreview:\n")
print(documents[0].text[:500])   # Show first 500 characters
```

```python
# ============================================================
# Cell 4: Chunk the Transcript
# ============================================================

# Create the sentence splitter
splitter = SentenceSplitter(
    chunk_size=300,
    chunk_overlap=50
)

# Split documents into nodes
nodes = splitter.get_nodes_from_documents(documents)

print(f"✅ Total Chunks Created: {len(nodes)}")

# Preview first 3 chunks
for i, node in enumerate(nodes[:3]):
    print(f"\n{'='*60}")
    print(f"Chunk {i+1}")
    print(f"{'='*60}")
    print(node.text[:500])
```

```python
# ============================================================
# Cell 5: Create Vector Index
# ============================================================

# Create the VectorStoreIndex
index = VectorStoreIndex(nodes)

print("✅ VectorStoreIndex created successfully!")

print(f"Total Nodes Indexed: {len(nodes)}")
```

```python
# ============================================================
# Cell 6: Create Dense Retriever
# ============================================================

# Create retriever
retriever = index.as_retriever(
    similarity_top_k=10
)

print("✅ Dense Retriever Created Successfully!")
print("Top-K Retrieval:", retriever.similarity_top_k)
```

```python
# ============================================================
# Cell 7: Test Dense Retrieval
# ============================================================

query = "Explain me networking"

# Retrieve relevant chunks
retrieved_nodes = retriever.retrieve(query)

print(f"Query: {query}")
print(f"\nRetrieved {len(retrieved_nodes)} Chunks\n")

for i, node in enumerate(retrieved_nodes):
    print("=" * 80)
    print(f"Rank {i+1}")
    print(f"Similarity Score: {node.score:.4f}")
    print("-" * 80)
    print(node.text)
    print()
```

```python
# ============================================================
# Cell 8: Build Zero-Shot Prompt
# ============================================================

query = "Summarize the entire video."

# Retrieve relevant chunks
retrieved_nodes = retriever.retrieve(query)

# Combine retrieved chunks into one context
context = "\n\n".join([node.text for node in retrieved_nodes])

# Zero-shot prompt
prompt = f"""
You are an expert AI assistant specialized in summarizing transcripts.

Use ONLY the information provided in the context below.
Do not invent or assume information that is not present.

-------------------- CONTEXT --------------------

{context}

-------------------------------------------------

Task:
Provide a structured summary including:

1. Main Topic
2. Key Concepts
3. Important Discussions
4. Conclusions
5. Key Takeaways

Keep the summary concise, informative, and under 300 words.
"""

print("="*80)
print(prompt)
print("="*80)
```

```python
# ============================================================
# Cell 9: Load LLM (Hugging Face Inference API)
# ============================================================
from google.colab import userdata
from huggingface_hub import InferenceClient
from getpass import getpass

# Enter your Hugging Face Access Token

# Initialize the client
client = InferenceClient(
    api_key=userdata.get('HF_TOKEN'),
)

print("✅ Hugging Face Client Initialized!")
```

```python
response = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[
        {"role": "user", "content": prompt}
    ],
    temperature=0.2,
    max_tokens=700,
    extra_body={
        "chat_template_kwargs": {
            "enable_thinking": False
        }
    }
)
```

```python
summary = response.choices[0].message.content
print(summary)
```
